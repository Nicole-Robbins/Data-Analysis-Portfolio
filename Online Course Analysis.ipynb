## The Data

I sourced this data from "Online Course Student Engagement Metrics" on Kaggle which was contributed by Yifan Zhu and can be found [here](https://www.kaggle.com/datasets/thedevastator/online-course-student-engagement-metrics). This dataset comes from the initial year of the Massive Open Online Courses that Harvard and MIT launched from Fall 2012 - Summer 2013. It contains data on student demographics, course enrollment, and engagement metrics. The university researchers published a study on their research from this if you wish to learn more!
Because these courses were a brand new style of learning and were open to anyone in the world regardless of enrollment at either university or other factors like age, ultimately the data doesn't point to an absolute correlation between demographics, metrics, and success. But what it can show is who might be interested in this learning format and did the certificate earners share any related behaviors.

This fields in this dataset include: 

- course id
- student id
- registered (signed up but never engaged with the content)
- viewed (looked at less than half of the content)
- explored (looked at half of the content or more)
- certified
- country
- education
- birth year
- final grade
- start date
- last event date
- number of events
- number of days active
- number of videos watched
- numbers of chapters read
- number of forum posts

## Data Cleaning

Because this was a pilot year for the brand new free online courses, there is a lot of extraneous data.
Here's some cleaning I did:

- Created calculated fields: Engagement Status, Age, Certified Status & Certified Rate, Explored Status & Explored Rate, Viewed Status & Viewed Rate, Course Duration
- Separated Course ID into 3 fields: Course Name, University, Course Semester

Data concerns: There are several date entry errors including some students being infants, some students finished the course before they actually started it, and some certificate earners had zero event activity which is impossible.

## Analyzing the Data

### Demographics

The demographics were broken down by age, gender, country, and education.

Over twice as many males registered as females, but females had a higher certification rate.



The majority of students were between 17-32, with the range reaching up to 80. Certification rates for the majority age group were typically between 2-3.5%, with 24-31 year olds having the highest certification rate. 31 year olds were at the top end with 3.14% and 18 year olds were outliers at a large drop down to 1.86%. 74 and 76 year olds had the highest certification rates overall(above 6%, and over 2% higher than the next highest age), but very low enrollment numbers.




The highest enrollment by current education level were those with Bachelor's degrees, followed by Secondary degrees. All education levels had certification rates between 2-4%, with Master's degree holders having the highest at 3.78% and Bachelor's degree holders at 2.37%.



By country, the highest enrollment came from the United States with just under 170k registrants, followed by India with less than half of that amount at around 78k registrants. Looking at certification rates, the US had an average rate at 2.57%, India was slightly above average with 3.93%, but some impressively high rates came from Spain at 8.71% and Poland at 8.14%.

 

### Most Common Student

The most common students were males between 22-30 who had a Bachelor's degree.



### Popular courses

Interestingly, while Harvard offered fewer classes it still had a far greater registration amount. There were just under 340k students who registered at Harvard, while around 250k enrolled at MIT. The top 5 courses were Harvard Intro to CompSci, MIT Intro to CompSci, Harvard Justice, Harvard Health Statistics, and Harvard Health and the Environment. Both universities offered an "Introduction to Computer Science" course, MIT even offered 2 of them, but Harvard's "Intro to CompSci" class had just under 170k registerants while both of MIT's combined totaled around 107k. A big driver behind this could be that Harvard has more name recognition, but it could also be that Harvard's CompSci course was the only Pass/Fail course.




Overall Harvard had a far higher enrollment rate, but MIT had a far greater certification rate!

### Behaviors of Certificate Earners

Course certification rates don't truly convey the success or impact of massive online open courses. There were large numbers of non-certified registrants who accessed substantial amounts of course content and could still have gained educational value without participating in assessments. However, should a student wish to earn certification in a course, it might be helpful for them to see the habits of other certificate earners.

While enrollment durations were similar between certificate earners and explorers, the average amount of interactive days for each group was very different. Both certificate earners and explorers were enrolled in the course for 140-150 days. Most certificate earners interacted with the course between 35-65 days, while explorers averaged between 15-35.



Interaction metrics showed very similar patterns of behavior across all engagement levels, however certificate earners on average had slightly higher levels of engagement than explorers did.




Grades of certificate earners across courses were between 73-93%, with the exception of Harvard IntroCompSci which was pass/fail.



## Key Takeaways

- The majority of students were from the United States, so broadening the course offerings and including translated material could garner more interest from students in other countries. Similarly, including more female instructors(especially at Harvard which had only male instructors) could encourage more female students to register. A very large number of students were particularly interested in Computer Science, so offering more courses in that field could lead to higher enrollment.
- Gathering more data on specific dates/times of interactions would be very beneficial to analyze. It could help predict when students could use more encouragement to stay engaged with the material, and help create a few options for suggested learning timelines for students who need to work at different paces. These things could be presented as introductory material at the start of the course that includes tips for completing the course and doing well on the assessments to help those students who wish to earn certification.
- Instructors could change the structures of their courses to better match with the trends of certificate earners such as keeping the number of chapters within the average range that the certificate earners read (10-20). With such large registration numbers, it wouldn't be possible for instructors to engage with every student individually, but it could be very helpful for the instructor to maintain steady engagement levels by posting encouragement or tips, or encouraging students to post questions in the forum and responding to them.
