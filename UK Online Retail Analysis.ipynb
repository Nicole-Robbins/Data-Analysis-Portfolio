## The Data

This data was sourced from the UC Irvine Machine Learning Repository, contributed by Daqing Chen and can be found [here](https://archive.ics.uci.edu/dataset/352/online+retail). It is a transactional data set which contains all the transactions occurring between Dec. 2010 and Dec. 2011 for a UK-based and registered non-store online retail.The company mainly sells unique all-occasion gifts, and many customers of the company are wholesalers.

This fields in this dataset include:
* Invoice Numbers
* Stock Codes
* Item Descriptions
* Quantity Purchased
* Unit Price
* Customer ID
* Country

## Data Cleaning

* created a calculated column called "InvoiceTotal" (Quantity * Unit Price)

Some concerns about the data: The net totals for each item might not be very accurate because returns were often input with a unit price of zero.

## Analyzing the Data

With an **Estimated Year-end Revenue of €9.3M** and **just under 5 million items sold**, this company had quite a lot of sales data to analyze.

I started by looking at the revenue streams at the highest level and breaking down which countries were buying the most products. Since this shop operates in the UK, the majority of sales naturally were from UK buyers.



UK buyers made up 85% of the store's yearly revenue!
With countries outside of the UK only making up 15% of the yearly revenue, there seems to be a great opportunity for the store owners to try and grow their market reach.



The highest spending countries were the Netherlands, EIRE, Germany, France, and Australia with a significant drop-off after that. 

Zooming in to look at what items performed the best, I broke it down into the best sellers(most sold) and highest grossing items.

 


Items 85123A(White Hanging Heart T-Light Holder) and 85099B(Jumbo Bag Red Retrospot) were in both categories. Despite being the highest grossing product, item 22423(Regency Cakestand 3 Tier) only sold 12,539 units, far below the range of the top sellers. The highest selling item 84077(World War 2 Gliders) sold for typically €0.29, and it actually only grossed roughly €13,200.

Each country had its own top selling items, but looking across all of the items there were clearly some international best sellers.



I also identified calendar trends broken down into best selling months and days of the week.



We can see that Thursdays typically perform the best. Interestingly, this store operates on Sundays but not Saturdays, which might be resulting in a huge missed opportunity. The daily sales trends were surprising to me because I was expecting Sunday and Monday to have the highest sales numbers, but I realized that I was thinking of consumer trends and this is a business that sells primarily to retailers.



From this chart, we can see that the autumn months see the most sales, with November clearly bringing in the highest sales numbers of the year. The monthly sales trends make sense because as a gift retailer, the months where sales are elevated are typically 2-4 weeks before a major holiday, with the autumn months leading to numerous back-to-back holidays.

## Key Takeaways

Without more information on customers, inventory, or website traffic, I focused my business suggestions on products, price, location, and date trends.
Here are my suggestions:

- With the UK making up 85% of the sales, the business should optimize their stock and  on prices for the top items from the UK- 84077, 22197, 85099B
- Avoid excess inventory for the worst performing items
- Sending out promotional content with new and popular seasonal items 6-4 weeks before holidays could help to capitalize on the popular sales windows
- While there is a very large gap between the UK revenue (85%) and the next top spending country of the Netherlands (3%), a slight increase in marketing to the top spending countries other than the UK (Netherlands, EIRE, Germany, France, and Australia) could still be impactful in the long-run
- For better data tracking, add another input field in the invoice process for "notes" because oftentimes for returns, adjustments, or issues the employee would write notes in the item description section. Add an input field for new or returning to keep track of what new customers are coming to the business to buy.
